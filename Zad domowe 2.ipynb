```python
from pyspark.sql import SparkSession

spark = (
    SparkSession.builder
    .appName("Lab2-Transactions")
    .getOrCreate()
)
spark.sparkContext.setLogLevel("WARN")
print(f"Spark {spark.version} — gotowy")
```

```python
df = spark.read.json("data/transactions_10k.jsonl")

print(f"Liczba rekordów: {df.count()}")
df.printSchema()
```

```python
df.show(10, truncate=False)
```

```python
from pyspark.sql.functions import to_timestamp, col

df = df.withColumn("timestamp", to_timestamp(col("timestamp"), "yyyy-MM-dd HH:mm:ss"))

df.printSchema()  # timestamp powinien być teraz 'timestamp (nullable = true)'
```

```python
from pyspark.sql.functions import count, sum as _sum, avg, round as _round

store_summary = (
    df.groupBy("store")
    .agg(
        count("tx_id").alias("liczba_tx"),
        _round(_sum("amount"), 2).alias("suma_PLN"),
        _round(avg("amount"), 2).alias("srednia_PLN"),
    )
    .orderBy("store")
)
store_summary.show()
```

```python
from pyspark.sql.functions import min as _min, max as _max

df.groupBy("category") \
  .agg(
      _sum("amount").alias("sum_amount"),
      _min("amount").alias("min_amount"),
      _max("amount").alias("max_amount")
  ) \
  .orderBy("category") \
  .show()
```

```python
from pyspark.sql.functions import window

hourly = (
    df.groupBy(window("timestamp", "1 hour"))    # okno 1-godzinne
    .agg(
        count("tx_id").alias("liczba_tx"),
        _round(_sum("amount"), 2).alias("suma_PLN"),
    )
    .orderBy("window")
)
hourly.show(truncate=False)
```

```python
(
    hourly
    .select(
        col("window.start").alias("od"),
        col("window.end").alias("do"),
        "liczba_tx",
        "suma_PLN",
    )
    .show(truncate=False)
)
```

```python
df.groupBy(
    window("timestamp", "30 minutes"),
    "store"
).agg(
    count("*").alias("transactions"),
    _sum("amount").alias("total_amount")
).orderBy(
    "window",
    "store"
).show(truncate=False)
```

```python
from pyspark.sql.functions import desc

df.filter(df.store == "Kraków") \
  .groupBy(window("timestamp", "1 hour")) \
  .agg(
      _sum("amount").alias("total_revenue")
  ) \
  .orderBy(desc("total_revenue")) \
  .show(truncate=False)
```

```python
sliding = (
    df.groupBy(window("timestamp", "1 hour", "30 minutes"))  # szerokość 1h, krok 30min
    .agg(
        count("tx_id").alias("liczba_tx"),
        _round(_sum("amount"), 2).alias("suma_PLN"),
    )
    .select(
        col("window.start").alias("od"),
        col("window.end").alias("do"),
        "liczba_tx",
        "suma_PLN",
    )
    .orderBy("od")
)
sliding.show(truncate=False)
```

```python
tumbling_rows = (
    df.groupBy(window("timestamp", "1 hour"))
    .agg(count("tx_id"))
    .count()
)
sliding_rows = (
    df.groupBy(window("timestamp", "1 hour", "30 minutes"))
    .agg(count("tx_id"))
    .count()
)
print(f"Tumbling (1h):          {tumbling_rows} okien")
print(f"Sliding  (1h / 30min):  {sliding_rows} okien")

# Odpowiedz w komentarzu: dlaczego sliding ma więcej wierszy?
# TWOJA ODPOWIEDŹ:
# Sliding window daje więcej wierszy, ponieważ jedno zdarzenie może należeć do wielu nakładających się okien. 
# W tumbling window okna się nie nakładają, więc każdy rekord należy tylko do jednego okna.
```

```python
# Odpowiedz na pytania w komentarzach:

# 1. Ile transakcji jest w oknie 09:00–10:00?
#    Sprawdź w wyniku zadania 3.1.
#    ODPOWIEDŹ: 4661

# 2. Jaka jest różnica między groupBy("store") a groupBy(window(...), "store")?
#    ODPOWIEDŹ:
# groupBy("store") grupuje dane tylko po sklepie, więc otrzymujemy jedną agregację dla każdego sklepu z całego zbioru danych.
# groupBy(window(...), "store") grupuje dane jednocześnie po przedziale czasowym i sklepie, dzięki temu dostajemy osobne wyniki dla każdego sklepu w każdym oknie czasowym.

# 3. W oknie sliding 1h/30min — ile okien zawiera transakcje z godziny 09:30?
#    ODPOWIEDŹ: dwa okna zawierają transakcje z godziny 09:30.
```

```python
# Praca domoowa
# 1. Znajdź godzinę, w której sklep Gdańsk miał najniższą średnią kwotę transakcji.
# ODPOWIEDŹ: Najniższa średnia była między 08:00 a 09:00

from pyspark.sql.functions import asc

df.filter(df.store == "Gdańsk") \
  .groupBy(window("timestamp", "1 hour")) \
  .agg(
      avg("amount").alias("avg_amount")
  ) \
  .orderBy(asc("avg_amount")) \
  .show(truncate=False)
```

```python
# 2. Policz ile transakcji per kategoria było w oknie 09:00–09:30.
# ODPOWIEDŹ: elektronika: 611, książki: 622, odzież: 605, żywność: 567 

df.filter(
    (col("timestamp") >= "2026-04-12 09:00:00") &
    (col("timestamp") < "2026-04-12 09:30:00")
).groupBy("category") \
 .count() \
 .orderBy("category") \
 .show()
```

```python
# 3. Zrób okno 15-minutowe i sprawdź w której ćwierćgodzinie był szczyt transakcji (łącznie dla wszystkich sklepów).
# ODPOWIEDŹ: Szczyt transakcji był w ćwierćgodzinie 09:15–09:30, liczba transakcji: 1234.

df.groupBy(
    window("timestamp", "15 minutes")
).agg(
    count("*").alias("transactions")
).orderBy(
    desc("transactions")
).show(1, truncate=False)
```
